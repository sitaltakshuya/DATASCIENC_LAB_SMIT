# LAB 6 : USE MATPLOTLIB LIBRARY TO VISUALIZE LINE, SCATTER , BAR CHART, HISTOGRAM , PIE, BOX/WISHKER PLOT

```python
#lINE PLOT
import matplotlib.pyplot as plt
import numpy as np 
xaxis=np.array([1,8])
yaxis=np.array([3,10])
plt.plot(xaxis, yaxis)
plt.show()
```

```python
# Draw a line with default x-axis starting with 0
ypoints=np.array([3,8,1,10,5,7])
#ypoint--> y axis data 
# ls---> line style
#lw----> line width
#ms-----> marker size
#mec----> marker edge color
#mfc-----> marker face color
plt.plot(ypoints,marker='*', ls="--",lw='5',ms=20,mec='g',mfc='r')
plt.plot()
```

```python
# Draw two line by specifying a plt.plot() for each line
import matplotlib.pyplot as plt
import numpy as np 
y1=np.array([3,8,1,10])
y2=np.array([6,2,1,11])
plt.plot(y1)
plt.plot(y2)
plt.show()
```

```python
# Draw two lines by specifying both x and y
x1=np.array([0,1,2,3])
y1=np.array([3,8,1,10])
x2=np.array([0,1,2,3])
y2=np.array([6,2,7,11])
plt.plot(x1,y1,x2,y2)
plt.show()
```

```python
x=np.array([80,85,90,95,100,105,110,115,120,125])
y=np.array([240,250,260,270,280,290,300,310,320,330])
# CAUTION ----> MAKE SURE THAT THE X AND Y ARE OF SAME SIZE 
plt.plot(x,y,'o:r',mfc='b')
plt.title("Excercise data",loc='center')
plt.xlabel("Duration")
plt.ylabel("Calories burnt")
plt.show()
```

```python
# Font Properties 
font1={'family':'serif','color':'blue','size':20}
font2={'family':'serif','color':'red','size':20}
plt.plot(x,y,'o:g',mfc='b')
plt.title("Sports Data New",fontdict=font1)
plt.xlabel("Hours",fontdict=font2)
plt.xlabel("Calories",fontdict=font2)
plt.grid()
plt.grid(axis='x')
plt.grid(axis='y')
plt.grid(color='green',linestyle="--",linewidth=0.5)
plt.show()
```

```python
# use subplot() to plot multiple plot on a sample plot
plt.subplot(2,1,1)
plt.plot(x1,y1)
plt.subplot(2,1,2)
plt.plot(x2,y2)
plt.show()
```

```python
# SCATTER PLOT 
# day one, the age and the speed of 13 cars
x1=np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y1=np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])

colors=np.random.randint(100,size=(13))
size=2*np.random.randint(100,size=(13))
plt.subplot(2,3,1)
# aplha-->controls the transp...
plt.scatter(x1,y1,c=colors,s=size,alpha=0.7,cmap="nipy_spectral")
# Day 2
x2=np.array([2,2,8,1,15,8,12,11,10,5,14,1,4])
y2=np.array([100,105,110,84,85,88,90,95,100,70,107,84,83])
plt.subplot(2,3,2)
plt.scatter(x2,y2,color='r')
## Multiple scatter plot in a single plot
plt.subplot(2,3,3)
plt.scatter(x1,y1,color='b')
plt.subplot(2,3,3)
plt.scatter(x2,y2,color='r')

# set own colors
colors=np.array(["red","green","blue","purple","beige","magenta","yellow","black","pink","orange","cyan","brown","gray"])
plt.subplot(2,3,4)
plt.scatter(x1,y1,c=colors)
# Create a color array and specify a colormap in the scatter plt
colors=np.array([0,10,20,30,40,45,50,55,60,70,80,90,100])
plt.subplot(2,3,5)
plt.scatter(x1,y1,c=colors,cmap='viridis')
```

```python
# Bar plot 
x=np.array(["A","B","C","D"])
y=np.array([3,8,1,10])
plt.subplot(2,1,1)
plt.bar(x,y,width=0.2)
plt.subplot(2,1,2)
plt.barh(x,y,color="red",height=0.5)
plt.show()
```

```python
# Histogram 
x=np.random.normal(170,10,250)
plt.hist(x)
plt.xlabel("Height of person in cm")
plt.ylabel("Number of people")
plt.show()
```

```python
# PIE CHART
y=np.array([35,25,25,15])
mylabels=["Apple","Banana","Kiwi","Blueberry"]
plt.subplot(3,2,1)
plt.pie(y,labels=mylabels)
plt.subplot(3,2,2)
plt.pie(y,labels=mylabels,startangle=90)
plt.subplot(3,2,3)
myexplode=[0,0,0,0.2]
plt.pie(y,labels=mylabels,explode=myexplode)
plt.subplot(3,2,4)
plt.pie(y,labels=mylabels,explode=myexplode,shadow=True)
plt.subplot(3,2,5)
plt.pie(y,labels=mylabels)
plt.legend(title="Four Fruits")
plt.show()
```

```python
# BOX PLOT 
# Create a data
data_1=np.random.normal(100,10,200)
data_2=np.random.normal(90,20,200)
data_3=np.random.normal(80,30,200)
data_4=np.random.normal(70,40,200)
data=[data_1,data_2,data_3,data_4]
fig=plt.figure(figsize=(10,7))
ax=fig.add_axes([0,0,1,1])
bp=ax.boxplot(data)
plt.show()
```
