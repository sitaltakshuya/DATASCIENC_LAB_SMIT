# LAB2: WRITE A PYTHON PROGRAM FOR COMPUTING MEAN, MEDIAN , MODE , STANDARD DEVIATION AND PERCENTILE

```python
#program to calculate the mean of an array using numpy.
import numpy as np
ar1=np.array([1,1,1,1,1,1,1,1,1])
mean=np.mean(ar1)
print("The original array is",ar1)
print(" The mean of the given array is:",mean)
```

```python
#program to calculate the median of an array 
#median=middle most value.
ar2=np.array([12,13,14,15,16,17,18,19,20])
med=np.median(ar2)
print("The original array is :",ar2)
print("The median value of the array is:",med)
```

```python
#Program to find the mode
#MODE: The most repetative value in an array.

import statistics as st
import numpy as np
ar3=np.array([10,20,20,30,40,50,60,10,10,100])
md=st.mode(ar3)
print("The original array is:", ar3)
print("\n")
print("The mode of the given array above is ",md)
```

```python
# Program to calculate the stabdard deviation of an array 
#S.D: How far a value is deviated from the mean 
# Higher the S.D: higher is the deviation
#Lower the S.D: Lower the deviation 
ar4=np.array([17,16,17,14,13,17])
sd=np.std(ar4)
m1=np.mean(ar4)
print("The original array is:",ar4)
print("\n")
print("The mean of the array is:",m1)
print("The calculated value of S.D is: ",sd)
```

```python
# Program to calcualte the percentile
per=np.array([10,14,15,17,45,50,55,60,80,100,33,63,72])
#Find out the 75 percentile of the data 
a=np.percentile(per,75)
print(a)
```
