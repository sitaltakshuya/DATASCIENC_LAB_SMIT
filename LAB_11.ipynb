```python
# reading the csv file
import pandas as pd
from pandas import read_csv
# plotting the graph
from matplotlib import pyplot
# Train-test split
from sklearn.model_selection import train_test_split
# Validation
from sklearn.model_selection import cross_val_score
# K -Fold 
from sklearn.model_selection import StratifiedKFold
# Importing logistic regression ----> sklearn
from sklearn.linear_model import LogisticRegression
# Importing KNN
from sklearn.neighbors import KNeighborsClassifier
```

```python
#url="https://github.com/jbrownlee/Datasets/master/iris.csv"
path="C:\\Users\\HP\\Downloads\\iris.csv"
names=['sepal-length','sepal-width','petal-length','petal-width','class']
dataset=read_csv(path,names=names)
#print(dataset)
# Split the data set into training, testting and validation
array=dataset.values
X=array[:,0:4]
y=array[:,4]
# We are creating a traing and validation set ---using---> train_test_split()
# test_size=0.20---> 20 % percent of data will be used as a test data
# random state=1 
# shuffle ---> shuffling the data
X_train, X_validation,Y_train,Y_validation=train_test_split(X,y,test_size=0.20,random_state=1,shuffle=True)
```

```python
# Empty list for the models 
models=[]
#models.append(('LR',LogisticRegression(solver='liblinear')))
models.append(('LR',LogisticRegression()))
models.append(('KNN',KNeighborsClassifier()))
print(models)
result=[]
names=[]
for name,model in models:
    kfold=StratifiedKFold(n_splits=10,random_state=1,shuffle=True)
    cv_results=cross_val_score(model,X_train,Y_train,cv=kfold,scoring='accuracy')
    result.append(cv_results)
    names.append(name)
    print('%s:%f(%f)'%(name,cv_results.mean(),cv_results.std()))
# Compare the algorithm 
pyplot.boxplot(result,label=names)
pyplot.title('Algorithm Comparision')
pyplot.show()
    
```
