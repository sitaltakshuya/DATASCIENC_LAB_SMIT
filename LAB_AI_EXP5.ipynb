```python
# Rules 
# it see and check every row and column and store the vaue at ...
# variable x
# go it dataframe(df) and check its index------> x---> row and col ( 1,1) 
for x in df.index:
    # check if in dataframe (df)--->loc(x) and rowname > 120:
    if df.loc[x,"Maxpulse"] > 120:
        # set the value as 120
        df.loc[x,"Maxpulse"] =120
    
print (df) 
```

```python
# Rule for Deletion 
for x in df.index:
    if df.loc[x,"Duration"]>120:
        df.drop(x, inplace=True)
```

```python
# Drop the duplicate value
# varaible name= dataframe(df)---> duplicated()---function---check---Duplicates-----return------> Boolean Value(T/F)
nv=df.duplicated()
#print(nv)
# Checking where the duplicates values lies??----Lies-----nv---becomes---True
new_out=df.loc[nv==True]
print(new_out)
# drop_duplicates----function----drop---Duplicate
df.drop_duplicates(inplace=True)
print(df.to_string())
```

```python
import pandas as pd
import numpy as np 
# Creating a dictionary--->{key:values}---key---F.S---Vlaue-->
dict={'First Score':[100,90,np.nan,95],
      'Second Score':[np.nan, 45,56,np.nan],
      'Third Score':[np.nan,40,80,95]
     }
df=pd.DataFrame(dict)
print(df.to_string())
```

```python
# CHECK IF THE VALUE IS NULL OR NOT
# isnull-->True--Value--->NULL
print(df.isnull())
```

```python
# check if the value  is not null
# notnull--->TRUE---->value--->NOTNULL
print(df.notnull())
```

```python
print(df)
```

```python
# Filling up the missing value 
print(df.replace(to_replace=np.nan,value=40))
```

```python
# Replace the NAN values with mean 
df_mean=df.mean()

print(df_mean)
# fillna---Function--->fill the NAN--->user choice of value
data_new=df.fillna(df.mean())
print(data_new)
```
