```python
# LINEAR REGRESSION ---> Y=MX+C
# DATA POINTS : X(INDEPENDENT) AND Y (DEPENDENT)
# DRAW A BEST FIT LINE 
# ! pip install scipy-----> slope , intercept , linear regression 
# Module not found scipy.stats or scipy
import scipy
import matplotlib.pyplot as plt
from scipy import stats
#Lab#aids
```

```python
# LINEAR REGRESSION = Y=Mx+C ---> M(slope), C(intercept), X(ind.var)
x=[5,7,8,7,2,17,2,9,4,11,12,9,6] # Independent Variable
y=[99,86,87,88,111,86,103,87,94,78,77,85,86] # Dependent variable
# X and Y are historical data (if x is 5 then y is 99)
```

```python
# calculate the slope , intercept , r, p , std_err

slope,intercept,r,p, std_err=stats.linregress(x,y)
def myfunc(x):
    return slope*x+intercept # ------> mx+c 
z=map(myfunc,x)# it works just like a loop map()
print(z)
mymodel=list(z)
print(mymodel)
plt.scatter(x,y)
plt.plot(x,mymodel)
plt.show()
```

```python
# Ploynomial regression 
# POLYNOMIAL REGRESSION 
import numpy 
x=[1,2,3,5,6,7,8,9,10,12,13,14,15,16,18,19,21,22]
y=[100,90,80,60,60,55,60,65,70,70,75,76,78,79,90,99,99,100]
```

```python
from sklearn.metrics import r2_score
p=numpy.polyfit(x,y,3) # least square polynomial fit
mymodel=numpy.poly1d(p)# 1 dimensional polynomial class
# Find---> r (squared value)---0 and 1 
# 0 means---> no relationship
# 1 means ---> perfect relationship
print(r2_score(y,mymodel(x))) # r squared value
myline=numpy.linspace(1,22,100)
plt.scatter(x,y)
plt.plot(myline,mymodel(myline))
plt.scatter(x,y)
plt.plot(myline,mymodel(myline))
plt.show()


```
